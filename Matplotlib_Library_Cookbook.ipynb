## Necessary Imports


```python
import matplotlib.pyplot as plt
import numpy as np
```

## Simple linear growth plot

```python
x = np.arange(0,10)
y = x
plt.plot(x,y)
plt.show()
```

## Simple exponential growth plot

```python
x = np.arange(0,10)
y = x**2
plt.plot(x,y)
plt.show()
```

---
## Linear, exponential and polynomial growth plot at the same graph
with x label, y label, grid and legend

```python
x = np.arange(0, 6)
y_linear = x
y_square = x**2
y_cube = x**3

plt.plot(x, y_linear, label='Linear Growth (y = x)')
plt.plot(x, y_square, label='Exponential Growth (y = x^2)')
plt.plot(x, y_cube, label='Polynomial Growth (y = x^3)')
plt.xlabel('This is X-axis')
plt.ylabel('This is Y-axis')
plt.legend() # Shows the plot lables in top left in the graph
plt.grid(True) # Shows grid (default is false)
plt.title("Linear vs Exponential vs Polynomial GROWTH") # For showing the lable of the whole graph.(Top middle)

plt.savefig('lin_exp_poly.png') # Save the graph
plt.show()

# If I use plt.savefig() after plet.show(), the plt.savefig() returns(or saves) a blank graph
```

## Subplot to plot 3 different plots in 3 different graphs but on one canvas.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10)
y_linear = x
y_square = x**2
y_cube = x**3

# Create a figure with two subplots
fig, axes = plt.subplots(1, 3, figsize=(16, 6)) # plt.subplots(<ROW>, <COLUMN>, figsize=(<WIDTH>, <HEIGHT>))

# Plot Linear Growth on the first subplot
axes[0].plot(x, y_linear, label='Linear Growth (y = x)', color='blue')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('Y-axis')
axes[0].set_title('Linear Growth')
axes[0].legend()
axes[0].grid(True)

# Plot Exponential Growth on the first subplot
axes[1].plot(x, y_square, label='Exponential Growth (y = x^2)', color='green')
axes[1].set_xlabel('X-axis')
axes[1].set_ylabel('Y-axis')
axes[1].set_title('Exponential Growth')
axes[1].legend()
axes[1].grid(True)

# Plot Polynomial Growth on the second subplot
axes[2].plot(x, y_cube, label='Polynomial Growth (y = x^3)', color='red')
axes[2].set_xlabel('X-axis')
axes[2].set_ylabel('Y-axis')
axes[2].set_title('Polynomial Growth')
axes[2].legend()
axes[2].grid(True)

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.suptitle('Comparison of Growth Patterns (Subplots)', y=1.02, fontsize=16) # Overall title
plt.show()
```

## Showing zoomed-in plot inside a plot using add_axes.
This is to show a plot inside a plot / show legend of a plot outside the plot / show a different plot inside or outside of a plot.

```python
# Data for the main plot
x = np.linspace(0, 10, 100)
y = x**2

# Create the main figure and axes
fig = plt.figure(figsize=(10, 8))
main_axes = fig.add_axes([0, 0, 1, 1]) # [left, bottom, width, height]

# Plot the main data
main_axes.plot(x, y, color='blue', label='y = x^2')
main_axes.set_xlabel('X-axis')
main_axes.set_ylabel('Y-axis')
main_axes.set_title('Exponential Growth with Zoomed Inset')
main_axes.grid(True)
main_axes.legend(loc='upper left')

# Create the inset axes for the zoomed-in plot
inset_axes = fig.add_axes([0.125, 0.625, 0.25, 0.25]) # [left, bottom, width, height] for the inset

# Plot the zoomed-in data
inset_axes.plot(x, y, color='red')
inset_axes.set_title('Zoomed-in (x=2 to 4)')
# Set explicit limits for the zoomed area
inset_axes.set_xlim(2, 4)
inset_axes.set_ylim(4, 16)
inset_axes.grid(True)

plt.show()
```
